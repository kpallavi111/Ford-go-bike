#### **Introduction**
This dataset contains structured information on **124,347 titles** from **Amazon Prime**, including **movies and TV shows**. With **19 attributes**, the dataset provides **detailed insights into title metadata, ratings, popularity, production details, and cast information**.

#### **Dataset Composition**
The dataset is formed by merging two CSV files:  
1️⃣ **titles.csv** – Includes general information like **title name, type, genres, description, age certification, release year, runtime, and ratings (IMDb & TMDb).**  
2️⃣ **credits.csv** – Contains **cast and crew details**, including their **roles, IMDb ID, and character names**.

#### **Key Attributes**
- **Title Details:** Movie/TV show **name, type (Movie or TV Show), description, and age certification**.  
- **Ratings & Popularity:** **IMDb score, IMDb votes, TMDb score, TMDb popularity.**  
- **Production Information:** **Release year, runtime, production countries, number of seasons.**  
- **Credits & Cast:** Details of actors, directors, and crew members, along with character names and roles.

#### **Data Types**
- **Object (11 columns)** – Text-based attributes like **title, genres, description, character names, and credits**.  
- **Integer (3 columns)** – Numeric values such as **release year, runtime, seasons**.  
- **Float (5 columns)** – Rating-based data including **IMDb score, IMDb votes, TMDb popularity, TMDb score, and season count**.  

#### **Data Quality Overview**
- **Duplicate Entries:** **168 records** are duplicated in the dataset.  
- **Missing Values:** Present in **9 key columns**, most notably:  
  - *Age Certification:* **67,640 missing**  
  - *Seasons:* **116,194 missing** (Mostly affecting movies, since TV shows have seasons)  
  - *IMDb ID:* **5,303 missing**  
  - *IMDb Score:* **6,051 missing**  
  - *IMDb Votes:* **6,075 missing**  
  - *Description:* **91 missing**

- **Audience Demographics Insights** – Assessing **age certifications, runtime preferences, and production trends**.

## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
```

```python
#remove dupliacted values
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#remove or replace null values
```

```python
df.isnull().sum()
```

```python
df['description'] = df['description'].fillna("Not Specified")
df['age_certification'] = df['age_certification'].fillna(df["age_certification"].mode()[0])
df['seasons'] = df['seasons'].fillna(df["seasons"].median())
df["imdb_id"]=df["imdb_id"].fillna("Not Available")
df["imdb_score"]=df["imdb_score"].fillna(df["imdb_score"].mean())
df["imdb_votes"]=df["imdb_votes"].fillna(df["imdb_votes"].mean())
df["tmdb_popularity"]=df["tmdb_popularity"].fillna(df["tmdb_popularity"].mean())
df["tmdb_score"]=df["tmdb_score"].fillna(df["tmdb_score"].mean())
df["character"]=df["character"].fillna("Not Disclosed")
```

### What all manipulations have you done and insights you found?

Throughout this project, various data manipulations were performed to clean and refine the Amazon Prime TV shows and movies dataset, allowing for meaningful analysis. IMDb Score vs. TMDb Score vs. Genre Popularity (Bubble Chart)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import ast # Import the ast module to safely evaluate the string literal

plt.figure(figsize=(10, 6))

# Convert the string representation of the list in 'genres' to an actual list
# and take the first genre as a simplified representation for hue.
# Handle potential errors if the string is not a valid list representation.
def get_first_genre(genre_list_str):
    try:
        genres = ast.literal_eval(genre_list_str)
        if isinstance(genres, list) and len(genres) > 0:
            return genres[0]
        else:
            return "Unknown"
    except:
        return "Unknown"

df['first_genre'] = df['genres'].apply(get_first_genre)


sns.scatterplot(x=df["imdb_score"], y=df["tmdb_score"], size=df["tmdb_popularity"], hue=df["first_genre"], alpha=0.6)
plt.title("IMDb Score vs. TMDb Score vs. Genre Popularity")
plt.xlabel("IMDb Score")
plt.ylabel("TMDb Score")
plt.legend(title="Genre", bbox_to_anchor=(1.05, 1), loc='upper left') # Adjust legend position
plt.tight_layout() # Adjust layout to prevent cutting off legend
plt.show()
```

##### 1. Why did you pick the specific chart?

This helps see how different genres perform based on IMDb and TMDb scores.



##### 2. What is/are the insight(s) found from the chart?

Shows whether certain genres tend to receive higher or lower ratings across platforms.
