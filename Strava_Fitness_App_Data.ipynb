## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

##Daily Activity

```python
# Chart - 1 visualization code
```

```python
# Convert date column
df1["ActivityDate"] = pd.to_datetime(df1["ActivityDate"])

# Plot
plt.figure(figsize=(12, 6))
sns.barplot(x="ActivityDate", y="TotalSteps", data=df1)
plt.xticks(rotation=45)
plt.title("Total Steps Per Day")
plt.show()
```

#### Chart - 2

## Daily Calories Line Chart

```python
# Chart - 2 visualization code
```

```python

df2["ActivityDay"] = pd.to_datetime(df2["ActivityDay"])

plt.figure(figsize=(12, 6))
plt.plot(df2["ActivityDay"], df2["Calories"], marker="o", linestyle="-")
plt.xlabel("Date")
plt.ylabel("Calories Burned")
plt.title("Calories Burned Over Time")
plt.xticks(rotation=45)
plt.show()
```

#### Chart - 3

## Daily Intensities Stacked Area Chart

```python
# Chart - 3 visualization code
```

```python
df3["ActivityDay"] = pd.to_datetime(df3["ActivityDay"])

df3.set_index("ActivityDay")[["VeryActiveMinutes", "FairlyActiveMinutes", "LightlyActiveMinutes", "SedentaryMinutes"]].plot.area(stacked=True, figsize=(12, 6))
plt.title("Activity Intensity Over Time")
plt.xlabel("Date")
plt.ylabel("Minutes")
plt.xticks(rotation=45)
plt.show()
```

#### Chart - 4

## Daily Steps Histogram

```python
# Chart - 4 visualization code
```

```python


plt.figure(figsize=(10, 5))
sns.histplot(df4["StepTotal"], bins=20, kde=True)
plt.title("Distribution of Daily Steps")
plt.xlabel("Step Total")
plt.ylabel("Frequency")
plt.show()
```

#### Chart - 5

## Heart Rate Box Plot

```python
# Chart - 5 visualization code
```

```python


plt.figure(figsize=(8, 5))
sns.boxplot(y=df5["Value"])
plt.title("Heart Rate Variability")
plt.ylabel("Heart Rate")
plt.show()
```

#### Chart - 6

## Sleep data PieChart

```python
# Chart - 6 visualization code
```

```python


plt.figure(figsize=(6, 6))
plt.pie([df6["TotalMinutesAsleep"].sum(), df6["TotalTimeInBed"].sum()], labels=["Asleep", "In Bed"], autopct="%1.1f%%", colors=["blue", "gray"])
plt.title("Sleep vs. Time in Bed")
plt.show()
```

# **Conclusion**

These visualizations provide an insightful overview of fitness habits, helping users understand their activity, calorie expenditure, sleep patterns, and heart rate trends.
